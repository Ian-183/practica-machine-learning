## Objetivo

Desarrollar un proceso completo de Machine Learning que construya un modelo de **regresión lineal múltiple** capaz de estimar los gastos médicos (`charges`) a partir de las características de cada persona (edad, sexo, BMI, dependientes, condición de fumador y región), cubriendo desde la carga de datos hasta la predicción de nuevas observaciones.

## Descripción del caso

Una organización de servicios de salud quiere saber **qué características se asocian con costos médicos más altos** y contar con un modelo que estime `charges` para personas nuevas. El dataset (Kaggle, *Medical Cost Personal Datasets*) tiene 1 338 registros y 7 columnas.

## Pipeline del proyecto (visión general)

El proyecto sigue **10 fases encadenadas**: la salida de cada una es la entrada de la siguiente. En cada fase se explica *qué se hace, con qué técnica y por qué*.

| # | Fase | Peso | Qué se hace | Técnica / herramienta | Salida |
|---|------|:---:|-------------|-----------------------|--------|
| 1 | Carga y comprensión | 10 % | Descarga automática desde Kaggle, lectura del CSV, inspección de estructura y diccionario de datos | `kagglehub`, `pandas` | `df` crudo |
| 2 | EDA y calidad | 15 % | Nulos, duplicados, tipos, distribuciones, relaciones con `charges`, correlaciones | `isna`, `duplicated`, histogramas, boxplots, `corr` | `df` limpio + hallazgos |
| 3 | Outliers | 10 % | Detectar atípicos y decidir qué hacer con ellos | Regla del IQR (Tukey) | Decisión justificada |
| 4 | Preparación de variables | 10 % | Separar `X` / `y`, partición *train/test* estratificada, codificar texto a números | `train_test_split`, `OneHotEncoder` | `X_train`, `X_test`, `y_train`, `y_test` |
| 5 | Normalización | 10 % | Poner las variables numéricas en la misma escala | `StandardScaler` (z-score) | `preprocesador` (ColumnTransformer) |
| 6 | Selección de atributos | 10 % | Medir cuánto aporta cada variable a explicar `charges` | `f_regression`, `mutual_info_regression`, `SelectKBest` | Ranking de variables |
| 7 | Validación | 10 % | Verificar integridad de los datos y elegir el nº de variables con validación cruzada | Chequeos con `assert`, `KFold`, `GridSearchCV` | `k` óptimo + métricas CV |
| 8 | Entrenamiento | 10 % | Ajustar el pipeline completo y leer los coeficientes | `Pipeline`, `LinearRegression` | `modelo_final` |
| 9 | Evaluación | 10 % | Medir el error sobre datos nunca vistos y revisar residuos | MAE, RMSE, MAPE, R², gráficas | Métricas en *test* |
| 10 | Predicción | 5 % | Estimar `charges` para personas nuevas | `modelo_final.predict` | Tabla de estimaciones |

**Regla de oro del pipeline (evitar fuga de datos):** todo lo que *aprende* de los datos (medias y desviaciones del escalador, categorías del codificador, puntajes de selección, coeficientes) se ajusta **solo con el conjunto de entrenamiento**. El conjunto de prueba se usa una única vez, al final, para evaluar.

```python
def obtener_dataset(slug="mirichoi0218/insurance", archivo="insurance.csv", destino="dataset"):
    """Descarga el dataset desde Kaggle (kagglehub). Si falla, usa una copia local."""
    os.makedirs(destino, exist_ok=True)
    ruta_destino = os.path.join(destino, archivo)
    try:
        try:
            import kagglehub
        except ImportError:
            subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "kagglehub"])
            import kagglehub
        carpeta = kagglehub.dataset_download(slug)           # descarga (o toma la caché de kagglehub)
        shutil.copy(os.path.join(carpeta, archivo), ruta_destino)
        print(f"Descargado desde Kaggle -> {ruta_destino}")
    except Exception as error:
        print(f"No se pudo descargar desde Kaggle ({type(error).__name__}). Se buscará una copia local.")
        for candidato in (ruta_destino, archivo):
            if os.path.exists(candidato):
                ruta_destino = candidato
                break
        else:
            raise FileNotFoundError("No hay copia local de insurance.csv.") from error
    return ruta_destino

ruta = obtener_dataset()
df = pd.read_csv(ruta)
print(f"Archivo en uso: {ruta}")
print(f"Dimensión: {df.shape[0]} filas x {df.shape[1]} columnas")
df.head()
```

# Fase 3. Detección de valores atípicos (10 %)

**Objetivo.** Identificar observaciones extremas y decidir con criterio si se eliminan, se transforman o se conservan.

**Procedimiento.** Se usa la **regla del rango intercuartílico (IQR) de Tukey**:
- `Q1` = percentil 25, `Q3` = percentil 75, `IQR = Q3 − Q1`.
- Límite inferior = `Q1 − 1.5·IQR`; límite superior = `Q3 + 1.5·IQR`.
- Todo valor fuera de esos límites se marca como atípico.

Después se revisa **quiénes** son los atípicos de `charges` (¿errores de captura o casos reales?).

**Justificación.** Eliminar un atípico es válido si es un error; si es un caso real, eliminarlo hace que el modelo ignore justo a las personas más costosas.

```python
def limites_iqr(serie, k=1.5):
    q1, q3 = serie.quantile([0.25, 0.75])
    iqr = q3 - q1
    return q1 - k * iqr, q3 + k * iqr

filas = []
for col in ["age", "bmi", "children", "charges"]:
    inf, sup = limites_iqr(df[col])
    mask = (df[col] < inf) | (df[col] > sup)
    filas.append({"variable": col, "límite inferior": inf, "límite superior": sup,
                  "atípicos": int(mask.sum()), "% del total": mask.mean() * 100})
tabla_outliers = pd.DataFrame(filas).set_index("variable")
tabla_outliers
```

```python
fig, ax = plt.subplots(1, 4, figsize=(15, 3.8))
for a, col in zip(ax, ["age", "bmi", "children", "charges"]):
    sns.boxplot(y=df[col], ax=a, color="#8da0cb")
    a.set_title(col)
plt.tight_layout(); plt.show()

# ¿Quiénes son los atípicos de charges y de bmi?
_, sup_ch = limites_iqr(df["charges"])
altos = df[df["charges"] > sup_ch]
print(f"charges por encima del límite ({sup_ch:,.0f}): {len(altos)}")
print(f"   de ellos, fumadores: {(altos['smoker'] == 'yes').sum()} ({(altos['smoker'] == 'yes').mean():.1%})")

inf_b, sup_b = limites_iqr(df["bmi"])
bmi_out = df[(df["bmi"] < inf_b) | (df["bmi"] > sup_b)]
print(f"\nbmi atípicos: {len(bmi_out)}  (BMI máximo = {bmi_out['bmi'].max():.1f}, valor extremo pero posible en la práctica)")
```

**Decisión: se conservan todos los atípicos.**
- `age` y `children` no tienen atípicos.
- Los atípicos de `bmi` son pocos y corresponden a valores extremos pero **posibles** (no hay BMI negativos ni imposibles).
- Los atípicos de `charges` son casi todos **fumadores**: no son errores de digitación sino el gasto real de ese grupo. Eliminarlos haría que el modelo subestime sistemáticamente a los fumadores, que son justamente los casos más costosos para la organización.

# Fase 4. Preparación de variables (10 %)

**Objetivo.** Dejar los datos listos para el modelo: separar entradas y objetivo, apartar un conjunto de prueba y convertir el texto en números.

**Procedimiento.**
1. **Separar `X` y `y`:** `y = charges` (lo que se quiere predecir) y `X` = las otras seis columnas. `charges` jamás puede quedar en `X`.
2. **Partición 80 % / 20 %** con `train_test_split(test_size=0.2, random_state=42)`. Se usa `stratify=smoker` para que fumadores y no fumadores tengan **la misma proporción** en entrenamiento y prueba (es la variable más influyente y está desbalanceada).
3. **Codificación de categóricas** con `OneHotEncoder(drop="first")`: cada categoría se vuelve una columna 0/1. Se elimina la primera categoría de cada variable para evitar la **trampa de la variable ficticia** (multicolinealidad perfecta: `smoker_yes` y `smoker_no` darían la misma información y la regresión lineal no podría estimar coeficientes únicos).

**Justificación.** La regresión lineal solo trabaja con números. Se particiona **antes** de ajustar cualquier transformación para que el conjunto de prueba no influya en nada del entrenamiento.

```python
COLS_NUM = ["age", "bmi", "children"]
COLS_CAT = ["sex", "smoker", "region"]

X = df.drop(columns="charges")
y = df["charges"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=SEED, stratify=X["smoker"]
)
print(f"X_train: {X_train.shape} | X_test: {X_test.shape}")
print("\nProporción de fumadores:")
print(pd.DataFrame({"train": X_train["smoker"].value_counts(normalize=True),
                    "test": X_test["smoker"].value_counts(normalize=True)}))
```

```python
# Demostración de la codificación (ajustada solo con train)
codificador_demo = OneHotEncoder(drop="first", sparse_output=False)
cat_codificadas = codificador_demo.fit_transform(X_train[COLS_CAT])
nombres_cat = codificador_demo.get_feature_names_out(COLS_CAT)

print("Columnas generadas:", list(nombres_cat))
print("Categoría eliminada (referencia) por variable:",
      {c: codificador_demo.categories_[i][0] for i, c in enumerate(COLS_CAT)})
pd.concat([X_train[COLS_CAT].head(4).reset_index(drop=True),
           pd.DataFrame(cat_codificadas[:4], columns=nombres_cat)], axis=1)
```

**Lectura.** Las 3 variables de texto se convierten en 5 columnas binarias: `sex_male`, `smoker_yes`, `region_northwest`, `region_southeast` y `region_southwest`. Las categorías `female`, `no` y `northeast` quedan como **referencia** (todas las columnas en 0).

---
# Fase 5. Normalización / estandarización (10 %)

**Objetivo.** Que `age`, `bmi` y `children` tengan escalas comparables.

**Procedimiento.** `StandardScaler` aplica la estandarización *z-score*:

$$ z = \frac{x - \mu}{\sigma} $$

donde `μ` y `σ` son la media y desviación estándar **del conjunto de entrenamiento**. Cada columna queda con media 0 y desviación 1. Las columnas 0/1 del one-hot **no** se escalan.

Después se arma el `ColumnTransformer` `preprocesador`, que aplica a cada grupo de columnas su tratamiento:
- **Numéricas:** `SimpleImputer(mean)` → `StandardScaler`.
- **Categóricas:** `SimpleImputer(most_frequent)` → `OneHotEncoder(drop="first")`.

Los imputadores no cambian nada con este archivo (0 nulos), pero hacen que el pipeline **no falle si una observación nueva llega con un campo vacío**.

**Justificación.** En regresión lineal pura la escala no cambia las predicciones, pero estandarizar permite **comparar la magnitud de los coeficientes** entre variables, es necesario para medir la importancia con métodos de selección y es buena práctica si luego se pasa a modelos con regularización (Ridge, Lasso).

```python
escalador_demo = StandardScaler().fit(X_train[COLS_NUM])
antes = X_train[COLS_NUM].agg(["mean", "std"]).T
despues = pd.DataFrame(escalador_demo.transform(X_train[COLS_NUM]), columns=COLS_NUM).agg(["mean", "std"]).T
print("ANTES (escala original) vs DESPUÉS (estandarizado):")
display(pd.concat({"antes": antes, "después": despues}, axis=1))
```

```python
preprocesador = ColumnTransformer(transformers=[
    ("num", Pipeline([("imputar", SimpleImputer(strategy="mean")),
                      ("escalar", StandardScaler())]), COLS_NUM),
    ("cat", Pipeline([("imputar", SimpleImputer(strategy="most_frequent")),
                      ("codificar", OneHotEncoder(drop="first", sparse_output=False,
                                                  handle_unknown="ignore"))]), COLS_CAT),
])

X_train_t = preprocesador.fit_transform(X_train)      # aprende SOLO con train
NOMBRES = preprocesador.get_feature_names_out()
print("Matriz transformada:", X_train_t.shape)
print("Columnas:", list(NOMBRES))
```

---
# Fase 6. Selección de atributos (10 %)

**Objetivo.** Saber cuáles variables aportan información real para explicar `charges` y cuáles solo meten ruido.

**Procedimiento.** Se calculan dos puntajes sobre el conjunto de **entrenamiento** transformado:
- **Estadístico F (`f_regression`)**: mide la relación *lineal* entre cada variable y `charges`; viene con un **p-valor** (si es < 0.05, la relación es estadísticamente significativa).
- **Información mutua (`mutual_info_regression`)**: mide dependencia de cualquier tipo (también no lineal). Sirve como contraste del F.

Luego `SelectKBest(f_regression, k)` conserva las `k` variables de mayor F. **El valor de `k` no se escoge a ojo**: se decide con validación cruzada en la Fase 7.

**Justificación.** Menos variables irrelevantes significa un modelo más simple, más fácil de explicar y menos propenso a sobreajustar.

```python
f_vals, p_vals = f_regression(X_train_t, y_train)
mi_vals = mutual_info_regression(X_train_t, y_train, random_state=SEED)

ranking = pd.DataFrame({"variable": NOMBRES, "F": f_vals, "p-valor": p_vals, "info. mutua": mi_vals})
ranking["significativa (p<0.05)"] = np.where(ranking["p-valor"] < 0.05, "sí", "no")
ranking = ranking.sort_values("F", ascending=False).reset_index(drop=True)
display(ranking.style.format({"F": "{:,.1f}", "p-valor": "{:.4f}", "info. mutua": "{:.3f}"}))

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.barplot(data=ranking, y="variable", x="F", ax=ax[0], color="#4c72b0"); ax[0].set_title("Estadístico F")
sns.barplot(data=ranking, y="variable", x="info. mutua", ax=ax[1], color="#dd8452"); ax[1].set_title("Información mutua")
plt.tight_layout(); plt.show()
```

**Lectura.** `smoker_yes` domina con diferencia, seguida de `age` y `bmi`. `children` aporta poco pero es significativa. Entre `sex_male` y las variables de región, solo `region_southeast` llega a ser significativa (p < 0.05) y con un F muy bajo; el resto no lo es. La decisión final de **cuántas** conservar se toma con validación cruzada (Fase 7), no con un corte arbitrario.

# Fase 7. Validación (10 %)

Esta fase tiene dos partes: **(A)** validar que los datos están bien preparados y **(B)** validar el modelo con validación cruzada para elegir `k`.

## 7A. Validación de la integridad de los datos

Se comprueban con `assert` las condiciones que el modelo da por supuestas. Si alguna falla, el notebook se detiene y avisa.

## 7B. Validación cruzada (K-Fold)

**Procedimiento.**
1. Se define el **pipeline completo** (`preprocesador → SelectKBest → LinearRegression`).
2. `KFold(n_splits=5, shuffle=True)` parte el *train* en 5 bloques. En cada vuelta se entrena con 4 y se mide con el bloque restante.
3. `GridSearchCV` repite esto para `k = 1 … 8` variables y compara el **RMSE promedio**.
4. Como el pipeline se vuelve a ajustar **dentro de cada vuelta**, el bloque de validación nunca participa en el escalado, la selección ni el ajuste de la recta: no hay fuga de datos.

**Justificación.** Un solo corte train/test puede dar un resultado bueno o malo por pura suerte. Con 5 cortes se obtiene un promedio y una desviación que indican qué tan estable es el modelo.

```python
# --- 7A. Chequeos de integridad ---
assert df.isna().sum().sum() == 0,                    "Hay valores nulos"
assert df.duplicated().sum() == 0,                    "Hay filas duplicadas"
assert "charges" not in X.columns,                    "charges quedó dentro de X (fuga del objetivo)"
assert df["age"].between(18, 64).all(),               "Edad fuera del rango esperado"
assert (df["bmi"] > 0).all() and (df["charges"] > 0).all(), "Valores no positivos en bmi/charges"
assert set(df["smoker"]) == {"yes", "no"},            "Categorías inesperadas en smoker"
assert set(X_train.index).isdisjoint(X_test.index),   "Hay filas en train y test a la vez"
assert list(X_train.columns) == list(X_test.columns), "train y test tienen columnas distintas"
assert len(X_train) + len(X_test) == len(df),         "Se perdieron filas en la partición"
print("Todos los chequeos de integridad pasaron correctamente.")
```

```python
# --- 7B. Validación cruzada y elección de k ---
pipeline_base = Pipeline([
    ("prepro", preprocesador),
    ("seleccion", SelectKBest(score_func=f_regression)),
    ("regresion", LinearRegression()),
])

cv = KFold(n_splits=5, shuffle=True, random_state=SEED)

busqueda = GridSearchCV(
    pipeline_base,
    param_grid={"seleccion__k": list(range(1, len(NOMBRES) + 1))},
    scoring={"rmse": "neg_root_mean_squared_error", "r2": "r2"},
    refit="rmse", cv=cv, return_train_score=False,
)
busqueda.fit(X_train, y_train)

res_k = pd.DataFrame({
    "k": busqueda.cv_results_["param_seleccion__k"].data.astype(int),
    "RMSE medio": -busqueda.cv_results_["mean_test_rmse"],
    "RMSE desv.": busqueda.cv_results_["std_test_rmse"],
    "R² medio": busqueda.cv_results_["mean_test_r2"],
})
display(res_k.style.format({"RMSE medio": "{:,.1f}", "RMSE desv.": "{:,.1f}", "R² medio": "{:.4f}"})
        .highlight_min(subset="RMSE medio", color="#c7e9c0"))

MEJOR_K = int(busqueda.best_params_["seleccion__k"])
print(f"\nMejor k según validación cruzada: {MEJOR_K}")

fig, ax = plt.subplots(figsize=(7, 3.8))
ax.errorbar(res_k["k"], res_k["RMSE medio"], yerr=res_k["RMSE desv."], marker="o", capsize=3)
ax.axvline(MEJOR_K, color="green", ls="--", label=f"k óptimo = {MEJOR_K}")
ax.set_xlabel("Número de variables conservadas (k)"); ax.set_ylabel("RMSE (validación cruzada)")
ax.set_title("Elección de k"); ax.legend(); plt.show()
```

```python
# Métricas de validación cruzada del modelo con el k elegido
pipeline_cv = clone(pipeline_base).set_params(seleccion__k=MEJOR_K)
scores = cross_validate(pipeline_cv, X_train, y_train, cv=cv,
                        scoring={"MAE": "neg_mean_absolute_error",
                                 "RMSE": "neg_root_mean_squared_error",
                                 "R²": "r2"})
resumen_cv = pd.DataFrame({
    "media": [-scores["test_MAE"].mean(), -scores["test_RMSE"].mean(), scores["test_R²"].mean()],
    "desviación": [scores["test_MAE"].std(), scores["test_RMSE"].std(), scores["test_R²"].std()],
}, index=["MAE", "RMSE", "R²"])
resumen_cv
```

**Lectura.** Con 1 o 2 variables el RMSE es claramente mayor (≈ 7 600 y ≈ 6 600); desde 3 variables el error se estabiliza alrededor de 6 300 y las mejoras posteriores son mínimas: entre `k = 4` y `k = 7` la diferencia es de apenas unos 40 USD, mucho menor que la desviación entre bloques (≈ 630 USD). La validación cruzada eligió `k = 7` por tener el RMSE promedio más bajo, pero la lectura honesta es que **sexo y región prácticamente no aportan**: si se prefiriera el modelo más simple, `k = 4` rendiría casi igual. En ambos casos el R² promedio es ≈ 0.72, con baja variación entre bloques, lo que indica un modelo **estable**.

# Fase 8. Entrenamiento mediante Regresión Lineal (10 %)

**Objetivo.** Ajustar el modelo final y poder explicar qué aprendió.

**Procedimiento.**
1. Se arma el pipeline definitivo con el `k` óptimo y se llama a `fit` con **todo el conjunto de entrenamiento**.
2. `LinearRegression` encuentra los coeficientes que minimizan la suma de errores al cuadrado (mínimos cuadrados ordinarios):

$$ \widehat{charges} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \dots + \beta_k x_k $$

3. Se leen los coeficientes en dos escalas: la **estandarizada** (para comparar importancia) y la **original** (para interpretar en dólares por unidad).

**Justificación.** En un modelo lineal los coeficientes son la explicación: cada uno dice cuánto cambia el gasto estimado al mover esa variable, manteniendo las demás constantes.

```python
modelo_final = clone(pipeline_base).set_params(seleccion__k=MEJOR_K)
modelo_final.fit(X_train, y_train)

prepro_ajustado = modelo_final.named_steps["prepro"]
mask = modelo_final.named_steps["seleccion"].get_support()
coefs = modelo_final.named_steps["regresion"].coef_
b0 = modelo_final.named_steps["regresion"].intercept_

escalador = prepro_ajustado.named_transformers_["num"].named_steps["escalar"]
desv = dict(zip(COLS_NUM, escalador.scale_))

filas = []
for nombre, c in zip(prepro_ajustado.get_feature_names_out()[mask], coefs):
    col = nombre.split("__")[1]
    c_orig = c / desv[col] if col in desv else c       # numéricas: se deshace la estandarización
    unidad = {"age": "por cada año", "bmi": "por cada punto de BMI", "children": "por cada hijo"}.get(col, "si la categoría es 'sí'")
    filas.append({"variable": nombre, "coef. estandarizado": c, "efecto en USD": c_orig, "interpretación": unidad})

tabla_coef = pd.DataFrame(filas).sort_values("coef. estandarizado", key=abs, ascending=False)
print(f"Variables que conserva el modelo (k = {MEJOR_K}) | intercepto = {b0:,.2f}")
display(tabla_coef.style.format({"coef. estandarizado": "{:,.2f}", "efecto en USD": "{:,.2f}"}))
```

**Lectura.** El modelo conserva 7 de las 8 columnas (descarta `region_northwest`). Manteniendo las demás variables constantes:
- **Fumar suma ≈ 23 750 USD** al gasto estimado, con diferencia el efecto más grande.
- **Cada año de edad suma ≈ 256 USD** y **cada punto de BMI ≈ 331 USD**.
- **Cada hijo suma ≈ 623 USD**.
- Vivir en el sureste o suroeste (frente al noreste) y ser hombre (frente a mujer) restan entre 260 y 1 020 USD: efectos pequeños y poco significativos, coherentes con lo visto en el EDA.

El intercepto (≈ 9 106 USD) es la estimación para una mujer no fumadora del noreste/noroeste con la edad, BMI y número de hijos promedio del entrenamiento.

# Fase 9. Evaluación del modelo (10 %)

**Objetivo.** Medir el error del modelo sobre **datos que nunca vio** (conjunto de prueba) y revisar dónde falla.

**Métricas.**
- **MAE:** error absoluto medio, en USD. Es el error "típico".
- **MSE / RMSE:** el error al cuadrado penaliza más los fallos grandes; el RMSE vuelve a USD.
- **MAPE:** error porcentual medio respecto al valor real.
- **R²:** proporción de la variación de `charges` que explica el modelo (1 = perfecto; 0 = no mejora a predecir siempre el promedio).

Se compara además contra un **modelo base** (`DummyRegressor`, que siempre predice la media del entrenamiento) para confirmar que el modelo realmente aprende algo.

**Diagnóstico gráfico.** *Real vs. predicho* y *residuos* (real − predicho) permiten ver si el error es aleatorio o tiene patrones que el modelo lineal no captura.

```python
y_pred = modelo_final.predict(X_test)

def calcular_metricas(y_real, y_est):
    mse = mean_squared_error(y_real, y_est)
    return {"MAE": mean_absolute_error(y_real, y_est), "MSE": mse, "RMSE": np.sqrt(mse),
            "MAPE (%)": np.mean(np.abs((y_real - y_est) / y_real)) * 100, "R²": r2_score(y_real, y_est)}

base = DummyRegressor(strategy="mean").fit(X_train, y_train)
comparacion = pd.DataFrame({
    "Modelo base (media)": calcular_metricas(y_test, base.predict(X_test)),
    "Regresión lineal (test)": calcular_metricas(y_test, y_pred),
})
display(comparacion)
print(f"R² promedio en validación cruzada (train): {resumen_cv.loc['R²', 'media']:.3f}  |  R² en test: {comparacion.loc['R²', 'Regresión lineal (test)']:.3f}")
```

```python
residuos = y_test - y_pred
fig, ax = plt.subplots(1, 3, figsize=(16, 4.3))

ax[0].scatter(y_test, y_pred, s=18, alpha=0.7, edgecolor="k", linewidth=0.3)
lim = [y.min(), y.max()]
ax[0].plot(lim, lim, "r--"); ax[0].set_xlabel("charges real"); ax[0].set_ylabel("charges predicho")
ax[0].set_title("Real vs. predicho")

ax[1].scatter(y_pred, residuos, s=18, alpha=0.7, edgecolor="k", linewidth=0.3)
ax[1].axhline(0, color="r", ls="--"); ax[1].set_xlabel("charges predicho"); ax[1].set_ylabel("residuo")
ax[1].set_title("Residuos vs. predicho")

sns.histplot(residuos, bins=30, kde=True, ax=ax[2], color="#4c72b0"); ax[2].set_title("Distribución de residuos"); ax[2].set_xlabel("residuo (real − predicho)")
plt.tight_layout(); plt.show()

# Error por grupo
por_grupo = pd.DataFrame({"smoker": X_test["smoker"], "real": y_test, "pred": y_pred})
por_grupo["error_abs"] = (por_grupo["real"] - por_grupo["pred"]).abs()
display(por_grupo.groupby("smoker").agg(personas=("real", "size"), gasto_medio=("real", "mean"), MAE=("error_abs", "mean")))
```

**Interpretación.**
- **Frente al modelo base:** el modelo explica ≈ 82 % de la variación de `charges` (R² = 0.82) y reduce el error absoluto medio de ≈ 9 081 USD a ≈ 3 605 USD. Hay un aprendizaje real, no un resultado por azar.
- **Sobreajuste:** el desempeño en *test* **no es peor** que en validación cruzada (R² 0.82 contra 0.72), así que no hay sobreajuste. Que sea mejor se debe en parte a que ese bloque de prueba resultó algo más fácil de predecir; por eso conviene reportar ambos valores y no solo el mejor.
- **MAPE ≈ 35 %:** parece alto porque los gastos pequeños (desde ≈ 1 100 USD) inflan el error porcentual; en dólares el error es moderado para la mayoría de personas.
- **Por grupos:** el MAE es ≈ 2 383 USD en no fumadores y ≈ 8 341 USD en fumadores. El modelo falla más justo en el grupo de mayor gasto.

**Límites del modelo lineal.** En el gráfico de residuos se ven agrupaciones y el error crece con el gasto estimado (varianza no constante). Revisando los residuos del *test*, los fumadores con BMI ≥ 30 quedan **subestimados** en promedio ≈ 7 400 USD y los fumadores con BMI < 30 **sobreestimados** ≈ 8 600 USD: el efecto de fumar *depende* del BMI (una interacción) y una recta sin interacciones no puede representarlo. Como mejoras futuras se proponen añadir el término `smoker × bmi`, modelar `log(charges)` o probar modelos no lineales.

# Fase 10. Predicción con nuevos datos (5 %)

**Objetivo.** Usar el modelo como lo usaría la organización: dar los datos de personas nuevas y obtener el gasto estimado.

**Procedimiento.** Se crea un DataFrame con las **mismas columnas que `X`**. No hay que escalar ni codificar a mano: `predict` recorre el pipeline y aplica lo ya aprendido en el entrenamiento. Se incluye una verificación de que cada persona esté dentro del rango de edad y BMI con el que se entrenó (fuera de ese rango, el modelo extrapola y la estimación es poco confiable).

El tercer y cuarto caso son la **misma persona** salvo por `smoker`, para aislar cuánto cambia la estimación solo por esa variable.

```python
nuevos = pd.DataFrame([
    {"age": 30, "sex": "female", "bmi": 25.4, "children": 0, "smoker": "no",  "region": "southwest"},
    {"age": 30, "sex": "male",   "bmi": 31.2, "children": 2, "smoker": "yes", "region": "southeast"},
    {"age": 45, "sex": "female", "bmi": 28.0, "children": 1, "smoker": "no",  "region": "northeast"},
    {"age": 45, "sex": "female", "bmi": 28.0, "children": 1, "smoker": "yes", "region": "northeast"},
])

nuevos["charges_estimado"] = modelo_final.predict(nuevos)

# Verificación de rango de entrenamiento
rango_ok = (nuevos["age"].between(X_train["age"].min(), X_train["age"].max()) &
            nuevos["bmi"].between(X_train["bmi"].min(), X_train["bmi"].max()))
nuevos["dentro_del_rango"] = np.where(rango_ok, "sí", "NO")

nuevos.style.format({"charges_estimado": "${:,.2f}"})
```

**Lectura.** Las dos últimas filas son idénticas salvo por `smoker`, y la diferencia entre sus estimaciones es el efecto aislado de fumar según el modelo. Las estimaciones son valores *esperados*, no cifras exactas: el error típico del modelo está dado por el MAE/RMSE de la Fase 9.

# Conclusión

El proyecto permitió recorrer, de principio a fin, el flujo completo de un problema de Machine Learning supervisado: descargar los datos de forma automática, revisar su calidad, decidir con criterio qué hacer con los valores atípicos, preparar y estandarizar las variables, seleccionar atributos, validar con validación cruzada, entrenar una regresión lineal múltiple, evaluarla con datos no vistos y usarla para estimar nuevos casos.

**Resultados principales.**
- El modelo final explica ≈ 82 % de la variación de los gastos médicos en el conjunto de prueba (R² = 0.82), con un error absoluto medio de ≈ 3 605 USD y un RMSE de ≈ 5 088 USD, frente a un MAE de ≈ 9 081 USD de un modelo que siempre predice el promedio.
- **Fumar es, con diferencia, el factor que más pesa:** suma ≈ 23 750 USD al gasto estimado. Le siguen la edad (≈ 256 USD por año), el BMI (≈ 331 USD por punto) y el número de hijos (≈ 623 USD por hijo). El sexo y la región aportan muy poco.
- Los gastos extremos no se eliminaron: casi el 98 % de ellos corresponde a fumadores, así que son casos reales y necesarios para que el modelo aprenda a estimar a quienes más cuestan.

**Limitaciones.** Una recta no capta que el efecto de fumar depende del BMI, por lo que el modelo subestima a los fumadores con BMI alto y sobreestima a los de BMI bajo, y su error es mayor justo en el grupo de mayor gasto. Además, el dataset tiene solo 1 337 registros y un único año, por lo que las estimaciones son orientativas y no deben aplicarse fuera del rango de edad (18–64) y BMI con el que se entrenó.

**Aprendizajes y reflexión.** Lo más valioso fue entender que el orden de las fases importa: particionar *antes* de escalar o seleccionar variables evita la fuga de datos, y encadenar los pasos en un `Pipeline` hace que ese cuidado se cumpla solo, también al validar y al predecir casos nuevos. También aprendí que una métrica sola no basta: el R² del *test* y el de la validación cruzada difirieron, y el análisis de residuos mostró límites que los números globales ocultaban. Como trabajo futuro se propone incluir la interacción `smoker × bmi`, transformar `charges` con logaritmo y comparar con modelos regularizados o de árboles. Esta actividad refuerza la formación en desarrollo de software al mostrar cómo convertir un conjunto de datos en un componente reproducible y explicable que puede integrarse en una aplicación real.

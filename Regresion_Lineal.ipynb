# REGRESIÓN LINEAL 

```python
# Importar las librerias
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Creamos el conjunto de datos
df = pd.DataFrame({
    "horas" : [1,2,3,4,5],
    "nota": [2,3,5,4,6]
})
X= df[["horas"]]
Y= df[["nota"]]
print (X)
```

```python
#Creamo nuestro modelo
modelo = LinearRegression()
modelo.fit(X, Y)
```

```python
#Imprimimos el interceptor y el coef
print("Interceptor: ", modelo.intercept_)
print("Pendiente: ", modelo.coef_[0])
```

```python
#Valores predichos
y_pred = modelo.predict(X)
```

```python
#Realizamos la nueva prediccion 7
horas=float(input("Ingrese las horas de estudio"))
prediccion = modelo.predict([[horas]])
print("Prediccion:", prediccion)
```

```python
plt.scatter(X, Y, color="blue", label="Datos reales")
plt.plot(X, y_pred, color="red", label="Línea de regresión")

# Agregamos etiquetas
plt.xlabel("Horas de estudio")
plt.ylabel("Nota")
plt.title("Relación entre horas de estudio y nota")

# Mostramos la leyenda
plt.legend()

# Mostramos la gráfica
plt.show()
```
